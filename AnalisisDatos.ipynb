## #11

```python
# Nulos antes de imputar
print(df_aire[["Material_Particulado_PM25", "Nivel_Ozono_ppb"]].isna().sum())

# PM2.5: reemplazar NaN por la media de la columna
media_pm25 = df_aire["Material_Particulado_PM25"].mean()
df_aire["Material_Particulado_PM25"] = df_aire["Material_Particulado_PM25"].fillna(media_pm25)

# Ozono: reemplazar NaN por la constante 0.0
df_aire["Nivel_Ozono_ppb"] = df_aire["Nivel_Ozono_ppb"].fillna(0.0)

# Nulos después de imputar 
print(df_aire[["Material_Particulado_PM25", "Nivel_Ozono_ppb"]].isna().sum())
```

## #12

```python
# Función que clasifica la temperatura
def clasificar_clima(temp):
    if temp < 18:
        return "Frío"
    elif temp <= 25:
        return "Templado"
    else:
        return "Cálido"

# Aplicarla con .apply() para crear la columna Clima
df_aire["Clima"] = df_aire["Temperatura_C"].apply(clasificar_clima)
df_aire[["Temperatura_C", "Clima"]].head()
```

## #13

```python
# Renombrar columnas
df_aire = df_aire.rename(columns={
    "Material_Particulado_PM25": "PM25",
    "Velocidad_Viento_kmh": "Viento_KMH"
})

# Reordenar: ID_Estacion y Clima primero
df_aire = df_aire[["ID_Estacion", "Clima", "Sector", "PM25", "Nivel_Ozono_ppb",
                    "Temperatura_C", "Humedad_Relativa", "Viento_KMH", "Alerta_Ambiental"]]

df_aire.head()
```

## #14

```python
# Agrupar por Sector: promedio de PM25 y máximo de Temperatura_C
df_aire.groupby("Sector").agg({"PM25": "mean", "Temperatura_C": "max"})
```

## #15

```python
# Función personalizada: rango = máximo - mínimo
def rango(series):
    return series.max() - series.min()

# Aplicarla con .agg() agrupando por Alerta_Ambiental
df_aire.groupby("Alerta_Ambiental")[["PM25", "Humedad_Relativa"]].agg(rango)
```

## #16

```python
# Agrupación multinivel: Sector y Alerta_Ambiental
df_aire.groupby(["Sector", "Alerta_Ambiental"]).agg(
    PM25_promedio=("PM25", "mean"),
    Humedad_promedio=("Humedad_Relativa", "mean")
)
```
